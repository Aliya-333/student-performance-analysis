
===============PERFORMANCE CATEGORY===============


```python
df["Performance"]=pd.cut(df["Final_Score"],bins=[0, 50, 70, 85, 100],
    labels=["Poor", "Average", "Good", "Excellent"]
)
```

```python
df["Performance"].value_counts()
```

```python
a = df["Performance"].value_counts()

plt.bar(a.index, a.values, color="black")

plt.title("Student Performance Distribution")
plt.xlabel("Performance")
plt.ylabel("Count")
plt.show()
```


============ Correlation Heatmap ⭐⭐⭐============


```python
numeric_columns = [
    "Age",
    "Study_Hours",
    "Attendance_Percentage",
    "Assignment_Score",
    "Midterm_Score",
    "Final_Score",
    "Previous_GPA",
    "Sleep_Hours"
]

correlation = df[numeric_columns].corr()
```

```python
plt.figure(figsize=(10,7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Between Student Performance Factors")
plt.show()
```

```python
df["Overall_Score"] = (
    df["Assignment_Score"] * 0.2 +
    df["Midterm_Score"] * 0.3 +
    df["Final_Score"] * 0.5
)
df.sort_values("Overall_Score", ascending=False).head(10)
```
